```python
items = [10, 20, 30, 40, 50, 60, 70]
print("First 3 elements : ", items[0:3])
print("Last 3 elements : ", items[-3:])
print("Every other elements : ", items[::2])
items.reverse()
print("Reversed list : ", items)
```

```python
scores = [85, 72, 90, 68, 95, 72, 80]
print("Maximum value :", max(scores), "& minimum value :", min(scores))
print("72 appears", scores.count(72), "times")
print("Index of 90 :",scores.index(90))
scores.reverse()
print("Reverse sorted list :", scores)
scores.sort()
print("In ascending order :", scores)
```

```python
classroom = {
"Alice": {"marks": 88, "grade": "A"},
"Bob": {"marks": 65, "grade": "B"},
"Carol": {"marks": 74, "grade": "B+"}
}
student=list(classroom.keys())
info="Name : "+student[0]+", Grade : "+ classroom["Alice"]["grade"]
print(info)
info="Name : "+ student[1] +", Grade : "+ classroom["Bob"]["grade"]
print(info)
info="Name : "+student[2]+", Grade : "+ classroom["Carol"]["grade"]
print(info)
classroom["Bob"]["marks"]=68
print("Updated Dictionary :", classroom)
```

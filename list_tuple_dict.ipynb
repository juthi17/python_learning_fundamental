```python
cart = [
{"name": "Rice", "price": 80, "qty": 5},
{"name": "Oil", "price": 180, "qty": 2},
{"name": "Sugar", "price": 90, "qty": 3},
{"name": "Butter", "price": 350, "qty": 1}
]
rice_cost=cart[0]["price"]*cart[0]["qty"]
print("Total cost of rice :",rice_cost,"BDT")
oil_cost=cart[1]["price"]*cart[1]["qty"]
print("Total cost of oil :",oil_cost,"BDT")
sugar_cost=cart[2]["price"]*cart[2]["qty"]
print("Total cost of sugar :",sugar_cost,"BDT")
butter_cost=cart[3]["price"]*cart[3]["qty"]
print("Total cost of butter :",butter_cost,"BDT")
print("Grand total of entire cart :", rice_cost + sugar_cost + oil_cost + butter_cost, "BDT")
```

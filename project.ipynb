```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=df
data.dtypes
```

```python
numeric_cols = ['dep_delay', 'arr_delay', 'distance', 'air_time']

for col in data.columns:
    if col in numeric_cols:
        next
    else:
        data[col] = data[col].astype(str)

data.dtypes
```

```python
data.isna().any()
```

```python
df.describe()
```

```python
data['distance'] = data['distance'] * 1.852
data.rename(columns={'distance': 'distance(km)'}, inplace=True)
data.columns
```

```python
bins = np.linspace(min(data['distance(km)']), max(data['distance(km)']), 4)
group_names = ['Short', 'Medium', 'Long']
data['distance_binned'] = pd.cut(data['distance(km)'], bins, labels = group_names, include_lowest = True)
print(data['distance_binned'])
```

```python
bins = np.linspace(min(data['air_time']), max(data['air_time']), 4)
group_names = ['Low', 'Medium', 'High']
data['air_time_binned'] = pd.cut(data['air_time'], bins, labels = group_names, include_lowest = True)
print(data['air_time_binned'])
```

```python
plt.hist(data['dep_delay'], bins=50)
plt.title('Distribution of Departure Delays')
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Number of Flights')
plt.show()
```

```python
plt.hist(data['arr_delay'], bins=50)
plt.title('Distribution of Arrival Delays')
plt.xlabel('Arrival Delay (minutes)')
plt.ylabel('Number of Flights')
plt.show()
```

```python
sns.countplot(data=data, x='name',
              order=data['name'].value_counts().index)
plt.title('Number of Flights by Airline')
plt.xlabel('Airline')
plt.ylabel('Number of Flights')
plt.xticks(rotation=90)
plt.show()
```

```python
avg_dep_delay = data.groupby('name')['dep_delay'].mean().sort_values()
avg_dep_delay.plot(kind='bar')
plt.title('Average Departure Delay by Airline')
plt.xlabel('Airline')
plt.ylabel('Average Departure Delay (minutes)')
plt.xticks(rotation=90)
plt.show()
```

```python
sns.countplot(data=data, x='origin')
plt.title('Number of Flights by Origin Airport')
plt.xlabel('Origin Airport')
plt.ylabel('Number of Flights')
plt.show()
```

```python
sns.countplot(data=data, x='name',
              order=data['name'].value_counts().index)
plt.title('Number of Flights by Airline')
plt.xlabel('Airline')
plt.ylabel('Number of Flights')
plt.xticks(rotation=90)
plt.show()
```

```python
avg_dep_delay = data.groupby('name')['dep_delay'].mean().sort_values()
avg_dep_delay.plot(kind='bar')
plt.title('Average Departure Delay by Airline')
plt.xlabel('Airline')
plt.ylabel('Average Departure Delay (minutes)')
plt.xticks(rotation=90)
plt.show()
```

```python
sns.countplot(data=data, x='origin')
plt.title('Number of Flights by Origin Airport')
plt.xlabel('Origin Airport')
plt.ylabel('Number of Flights')
plt.show()
```

```python
plt.scatter(data['distance(km)'], data['air_time'])
plt.title('Relationship Between Distance and Air Time')
plt.xlabel('Distance (km)')
plt.ylabel('Air Time (minutes)')
plt.show()
```

```python
monthly_flights = data.groupby('month')['flight'].count()
plt.plot(
    monthly_flights.index,
    monthly_flights.values,
    marker='o')
plt.title('Number of Flights by Month')
plt.xlabel('Month')
plt.ylabel('Number of Flights')
plt.xticks(range(1, 13))
plt.show()
```

```python
heatmap_data = data.pivot_table(
    		values='air_time',
    		index='month',
    		columns='origin',
    		aggfunc='mean')
heatmap_data = heatmap_data.reindex(range(1, 13))
sns.heatmap(heatmap_data, fmt='.1f')
plt.title('Average Air Time by Month and Origin Airport')
plt.xlabel('Origin Airport')
plt.ylabel('Month')
plt.show()
```

```python
origin_counts = data['origin'].value_counts()
plt.pie(
    origin_counts,
    labels=origin_counts.index,
    autopct='%1.1f%%')
plt.title('Percentage of Flights by Origin Airport')
plt.show()
```

```python
sns.boxplot(
    data=data,
    x='origin',
    y='arr_delay'
)

plt.title('Arrival Delay by Origin Airport')
plt.xlabel('Origin Airport')
plt.ylabel('Arrival Delay (minutes)')
plt.show()
```

```python
r=5.5
h=12.0
pi=3.14159265
base_area=2*pi*r**2
lateral=2*pi*r*h
Total_surface=2*base_area+lateral
volume=base_area*h
base_area=round(base_area,4)
lateral=round(lateral,4)
Total_surface=round(Total_surface,4)
volume=round(volume,4)
print("Base area ", base_area)
print("Lateral(side) surface area ", lateral)
print("Total surface area ", Total_surface)
print("Volume ", volume)
```

```python
usd=1500.75
rate_bdt=110.5
rate_eur=0.92
rate_gbp=0.79
rate_usd_bdt=usd*rate_bdt
rate_usd_bdt=round(rate_usd_bdt,2)
print("1500.75 USD =",rate_usd_bdt,"BDT")
rate_usd_eur=usd*rate_eur
rate_usd_eur=round(rate_usd_eur,2)
print("1500.75 USD =", rate_usd_eur," EUR")
rate_usd_gbp=usd*rate_gbp
rate_usd_gbp=round(rate_usd_gbp,2)
print("1500.75 USD =", rate_usd_gbp,"GBP")
usd_from_bdt=5000/rate_bdt
usd_from_bdt=round(usd_from_bdt,2)
eur_from_usd=usd_from_bdt*rate_eur
eur_from_usd=round(eur_from_usd,2)
gbp_from_usd=usd_from_bdt*rate_gbp
gbp_from_usd=round(gbp_from_usd,2)
print("5000 BDT = ", usd_from_bdt,"USD = ",eur_from_usd," EUR = ", gbp_from_usd, " GBP")
```
